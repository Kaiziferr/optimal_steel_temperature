**Data steel**

```python
data_steel[[
    'active_power',
    'reactive_power'
]] = data_steel[[
    'active_power',
    'reactive_power'
]].astype("float64")

data_steel['arc_heating_start'] = pd.to_datetime(data_steel['arc_heating_start'])
data_steel['arc_heating_end'] = pd.to_datetime(data_steel['arc_heating_end'])
```

```python
temp = np.datetime64(data_steel['arc_heating_end'][1]) - np.datetime64(data_steel['arc_heating_start'][1])
print(temp.astype('timedelta64[m]'))
print(temp / np.timedelta64(1, 'm'))
```

```python
# Calculate minutes
data_steel['minutes_process'] = data_steel.apply(lambda x: tools.calculate_time(x, 'arc_heating_start', 'arc_heating_end'), axis=1)
```

```python
# Get month
data_steel['months'] = data_steel['arc_heating_start'].dt.month
```

```python
data_steel.head()
```

```python
function_dictionary = {
    'active_power': ['sum'],
    'reactive_power': ['sum'],
    'minutes_process': ['sum']
}

new_val_col = [
    'active_power_sum',
    'reactive_power_sum',
    'minutes_process_sum'
]


data_group = data_steel.groupby([
    'key',
    'months'
]).aggregate(
    function_dictionary
)

data_group.columns = new_val_col
data_group = data_group.reset_index()
```

```python
data_group.head()
```

```python
data_group[data_group.duplicated(subset=['key'], keep=False)]
```

This rows explain that the steel arcs were manufactured in more two month, which is seems like a mistake. the rows with less minutes of manufactured will deleted

```python
data_group = data_group.sort_values(['minutes_process_sum'], ascending=False) 
data_group = data_group.drop_duplicates('key')
```

```python
function_dictionary = {
    'active_power_sum': ['sum'],
    'reactive_power_sum': ['sum'],
    'minutes_process_sum': ['sum'],
    'months': ['count']
}

new_val_col = [
    'active_power_sum',
    'reactive_power_sum',
    'minutes_process_sum',
    'active_months'
]


data_group = data_group.groupby([
    'key',
]).aggregate(
    function_dictionary
)

data_group.columns = new_val_col
data_group = data_group.reset_index()
```
