```python
import warnings

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import (
    train_test_split
)

from sklearn.metrics import mean_absolute_error

from lazypredict.Supervised import LazyRegressor


from sklearn.ensemble import (
    GradientBoostingRegressor,
    AdaBoostRegressor,
    HistGradientBoostingRegressor,
    RandomForestRegressor)

from sklearn.linear_model import (
    LinearRegression,
    Ridge,
    ElasticNet)

from xgboost import XGBRegressor, XGBRFRegressor

from sklearn.dummy import DummyRegressor
```

# **Config**
---

```python
random_seed = 12354
sns.set(style="darkgrid")
sns.set(font_scale=1)
pd.set_option('display.float_format', '{:,.5f}'.format)
title_data = 'Temperature prediction'
paleta = sns.color_palette("tab10").as_hex()
warnings.filterwarnings("ignore")
```

To validate the initial behavior of the models that best fit, a wrapper will be used that groups default instances of regression models, which can provide an indication of how the models behave in response to data from different strategies.

LazyRegressor is used, with the mean_absolute_error metric

 However, it will be limited to the following models:

- GradientBoostingRegressor
- AdaBoostRegressor,
- HistGradientBoostingRegressor,
- RandomForestRegressor,
- LinearRegression,
- Ridge,
- ElasticNet,
- XGBRegressor,
- XGBRFRegressor
- DummyRegressor


Since LazyRegressor provides a number of models, not all of which are well-known.

```python
wra_regressor = LazyRegressor(
    verbose=0,
    ignore_warnings=True, 
    custom_metric=mean_absolute_error, 
    predictions=True, 
    regressors=[
        GradientBoostingRegressor,
        AdaBoostRegressor,
        HistGradientBoostingRegressor,
        RandomForestRegressor,
        LinearRegression,
        Ridge,
        ElasticNet,
        XGBRegressor,
        XGBRFRegressor,
        DummyRegressor
        ])
```

# **Data First Strategy**
---

The first strategy is loaded.

```python
data_first = pd.read_csv('../data/first_strategy.csv')
```

**Split**

```python
X = data_first[[
    'active_power_sum', 'reactive_power_sum', 'minutes_process_sum',
    'gas_purgue'
]]

y = data_first['temperature_sum']
```

Randomly split the data into training and validation sets.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    train_size=0.85,
    shuffle=True,
    random_state=random_seed
)
```

The container is fitted with the training and validation data to adjust all the regression models offered by LazyRegressor.

```python
regressor_model_first_strategy , predictions  = wra_regressor.fit(
    X_train,
    X_test,
    y_train,
    y_test
)
```

```python
regressor_model_first_strategy.sort_values('mean_absolute_error')
```

The models typically used for linear problems have poorer performance, and their error difference is significant compared to ensemble models.
